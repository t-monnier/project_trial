```python
# TODO: Add docstring and explanatory comments
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/bcn_listings.csv.gz")
# TODO: Add a second dataset
```

```python
# Clean price
df["price"] = df["price"].str.replace("$", "", regex=False)
df["price"] = df["price"].str.replace(",", "", regex=False)
df["price"] = df["price"].astype(float)
# TODO: Clean one variable from the second dataset
```

```python
# Subset: entire apt accommodates 2 to 6 people from selected areas
areas = ["Eixample", "Ciutat Vella", "Gràcia", "Sarrià-Sant Gervasi"]
myquery = (
 "room_type == 'Entire home/apt' "
 "and 1 < accommodates < 7 and "
 "neighbourhood_group_cleansed in @areas")
# Create interpretable variable
df["price_person"] = df["price"] / df["accommodates"]
```

```python
# Analysis by neighbourhood and number of guests
price_pivot = pd.pivot_table(
 df.query(myquery),
 index="accommodates", # ["accommodates", "another", "other"]
 columns="neighbourhood_group_cleansed",
 values="price_person",
 aggfunc="median")
price_pivot
```

```python
# TODO: Ensure the graph includes clear title, labels and legend
price_pivot.plot( kind="line",
 marker="o",
 title="Median Price per Person")
# Save the graph to the reports/ folder and display it
plt.savefig("../reports/graph3.png")
plt.show()
```
